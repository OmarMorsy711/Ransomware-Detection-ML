```python
df = df[df["Category"].str.lower().isin(["benign", "ransomware"])]

benign = df[df["Category"].str.lower() == "benign"]
ransom = df[df["Category"].str.lower() == "ransomware"]

print("benign rows:", len(benign))
print("ransomware rows:", len(ransom))
```

**T2: Process Data**

```python
benign = df[df["Category"] == "Benign"]
ransom = df[df["Category"] == "Ransomware"]

print(len(benign), len(ransom))


benign_sampled = benign.sample(n=5000, random_state=42)


df_balanced = pd.concat([benign_sampled, ransom], axis=0)


df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)


df_balanced.to_csv("ransom_balanced.csv", index=False)
```

```python
df = pd.read_csv('ransom_balanced.csv')
```

```python
for col in df.columns:
    print(col, "=>", df[col].nunique())
```

```python
print(len(df))
```

```python
df = df.drop(columns=["sha1", "Family", "Class", "magic_number","md5", "network_threats", "file_extension"], inplace=False)
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
```

```python
cats = df.select_dtypes(include='object').columns
nums = df.select_dtypes(include='float64').columns

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop = "if_binary",handle_unknown='ignore', sparse_output=False), cats),
        ('num', StandardScaler(), nums)
    ], remainder='drop')
```

```python
data_processed = preprocessor.fit_transform(df)
encoded_columns = preprocessor.transformers_[0][1].get_feature_names_out(cats)
scaled_columns = preprocessor.transformers_[1][2]
all_columns = list(encoded_columns) + list(scaled_columns)
```

```python
data_final= pd.DataFrame(data_processed, columns=all_columns)
data_final.head()
```

```python
display(data_final['Category_Ransomware'].value_counts())
```

**T3: Partition Data**

```python
from sklearn.model_selection import train_test_split
```

```python
X_data= data_final.drop(columns=['Category_Ransomware'])
y_lable= data_final['Category_Ransomware']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_data, y_lable, test_size=0.3, random_state=42, stratify=y_lable, shuffle=True)
```

```python
print(f"training data:{X_train.shape}, test data:{X_test.shape}")
```

```python
display(y_train.value_counts())
display(y_test.value_counts())
```

**T4: Create Models**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform
```

```python
lg = LogisticRegression()

lg_params = {
    'C': uniform(0.01, 10),
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear','lbfgs']
}
```

```python
lg_search = RandomizedSearchCV(lg, lg_params,
                              n_iter=10, cv=5, random_state=42,
                              scoring='balanced_accuracy',n_jobs=-1)

lg_search.fit(X_train, y_train)
```

```python
lg_search.best_score_
```

```python
lg_fitted= lg_search.best_estimator_
print(lg_fitted)
```

**T5 Evaluate Models**

```python
lg_y_pred_test =  lg_fitted.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import balanced_accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
def cm(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()

    print("Confusion Matrix (rows = true, columns = predicted)")
    print("            Pred 0 (Benign)   Pred 1 (Ransomware)")
    print(f"True 0  (Benign)    TN = {tn}          FP = {fp}")
    print(f"True 1  (Ransom)    FN = {fn}          TP = {tp}")
    print()
    print(f"TN = {tn}, FP = {fp}, FN = {fn}, TP = {tp}")
```

```python
eval_test=[]

def evaluate(y_test,y_test_pred):
  accuracy_test=accuracy_score(y_test,y_test_pred)

  bal_accuracy_test=balanced_accuracy_score(y_test,y_test_pred)

  precision_test=precision_score(y_test,y_test_pred)

  recall_test=recall_score(y_test,y_test_pred)

  f1_test=f1_score(y_test,y_test_pred)

  eval_test.append({"accuracy": accuracy_test, "balanced acc": bal_accuracy_test,\
                    "precision": precision_test, "recall": recall_test,
                    "f1": f1_test
                    })
```

```python
evaluate(y_test,lg_y_pred_test)
```

```python
test_eval_all=pd.DataFrame(eval_test,index=['lg'])
```
